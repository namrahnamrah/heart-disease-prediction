```python
import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report,
    roc_curve
)

# =========================================================
# 1. SETUP
# =========================================================
sns.set_style("whitegrid")
os.makedirs("outputs", exist_ok=True)

# =========================================================
# 2. LOAD DATA
# =========================================================
df = pd.read_csv("heart_disease_clean.csv")

print("Original shape:", df.shape)
print("Missing values:\n", df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())

# =========================================================
# 3. CLEAN DATA
# =========================================================
df = df.drop_duplicates().copy()
print("\nShape after dropping duplicates:", df.shape)

# leakage check
print("\nCross-tab of num vs target:")
print(pd.crosstab(df["num"], df["target"]))

# remove leakage column
X = df.drop(columns=["target", "num"])
y = df["target"]

print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)
print("\nFeatures:")
print(X.columns.tolist())

# =========================================================
# 4. EDA PLOTS
# =========================================================
# 4.1 Target distribution
plt.figure(figsize=(6, 4))
sns.countplot(x=y)
plt.title("Target Class Distribution")
plt.xlabel("Heart Disease Target")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig("outputs/target_distribution.png", dpi=300)
plt.show()

# 4.2 Numeric histograms
numeric_cols = ["age", "trestbps", "chol", "thalch", "oldpeak"]

df[numeric_cols].hist(figsize=(12, 8), bins=20)
plt.suptitle("Distributions of Numeric Features", y=1.02)
plt.tight_layout()
plt.savefig("outputs/numeric_histograms.png", dpi=300)
plt.show()

# 4.3 Boxplots by target
plt.figure(figsize=(14, 10))
for i, col in enumerate(numeric_cols, 1):
    plt.subplot(3, 2, i)
    sns.boxplot(x="target", y=col, data=df)
    plt.title(f"{col} by Target")
plt.tight_layout()
plt.savefig("outputs/boxplots_by_target.png", dpi=300)
plt.show()

# 4.4 Correlation heatmap
plt.figure(figsize=(16, 12))
sns.heatmap(df.corr(numeric_only=True), cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("outputs/correlation_heatmap.png", dpi=300)
plt.show()

# =========================================================
# 5. TRAIN / TEST SPLIT
# =========================================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# scale for SVM, PCA, MLP
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_scaled_full = scaler.fit_transform(X)

# =========================================================
# 6. PCA (UNSUPERVISED METHOD)
# =========================================================
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled_full)

pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_df["target"] = y.values

plt.figure(figsize=(8, 6))
sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="target")
plt.title("PCA of Heart Disease Dataset")
plt.tight_layout()
plt.savefig("outputs/pca_plot.png", dpi=300)
plt.show()

print("\nPCA explained variance ratio:", pca.explained_variance_ratio_)
print("Total explained variance:", pca.explained_variance_ratio_.sum())

# =========================================================
# 7. MODEL 1: SVM
# =========================================================
svm = SVC(kernel="rbf", probability=True, random_state=42)
svm.fit(X_train_scaled, y_train)

svm_preds = svm.predict(X_test_scaled)
svm_probs = svm.predict_proba(X_test_scaled)[:, 1]

# =========================================================
# 8. MODEL 2: MLP NEURAL NETWORK
# =========================================================
mlp = MLPClassifier(
    hidden_layer_sizes=(32, 16),
    activation="relu",
    solver="adam",
    max_iter=500,
    random_state=42
)
mlp.fit(X_train_scaled, y_train)

mlp_preds = mlp.predict(X_test_scaled)
mlp_probs = mlp.predict_proba(X_test_scaled)[:, 1]

# =========================================================
# 9. EVALUATION
# =========================================================
def evaluate_model(name, y_true, y_pred, y_prob):
    return {
        "Model": name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred),
        "Recall": recall_score(y_true, y_pred),
        "F1 Score": f1_score(y_true, y_pred),
        "ROC-AUC": roc_auc_score(y_true, y_prob)
    }

svm_results = evaluate_model("SVM", y_test, svm_preds, svm_probs)
mlp_results = evaluate_model("MLP Neural Network", y_test, mlp_preds, mlp_probs)

results_df = pd.DataFrame([svm_results, mlp_results])
print("\nModel comparison:")
print(results_df)

results_df.to_csv("outputs/model_comparison_results.csv", index=False)

# =========================================================
# 10. CONFUSION MATRICES
# =========================================================
for model_name, preds in [
    ("SVM", svm_preds),
    ("MLP Neural Network", mlp_preds)
]:
    cm = confusion_matrix(y_test, preds)
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.title(f"Confusion Matrix - {model_name}")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    filename = model_name.lower().replace(" ", "_") + "_confusion_matrix.png"
    plt.savefig(f"outputs/{filename}", dpi=300)
    plt.show()

# =========================================================
# 11. ROC CURVES
# =========================================================
svm_fpr, svm_tpr, _ = roc_curve(y_test, svm_probs)
mlp_fpr, mlp_tpr, _ = roc_curve(y_test, mlp_probs)

plt.figure(figsize=(7, 5))
plt.plot(svm_fpr, svm_tpr, label=f"SVM (AUC = {roc_auc_score(y_test, svm_probs):.3f})")
plt.plot(mlp_fpr, mlp_tpr, label=f"MLP (AUC = {roc_auc_score(y_test, mlp_probs):.3f})")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()
plt.tight_layout()
plt.savefig("outputs/roc_curve_comparison.png", dpi=300)
plt.show()

# =========================================================
# 12. CLASSIFICATION REPORTS
# =========================================================
svm_report = classification_report(y_test, svm_preds)
mlp_report = classification_report(y_test, mlp_preds)

print("\nSVM Classification Report:\n")
print(svm_report)

print("\nMLP Classification Report:\n")
print(mlp_report)

with open("outputs/svm_classification_report.txt", "w") as f:
    f.write(svm_report)

with open("outputs/mlp_classification_report.txt", "w") as f:
    f.write(mlp_report)

# =========================================================
# 13. SAVE SUMMARY TEXT
# =========================================================
with open("outputs/project_summary.txt", "w") as f:
    f.write("Predictive Data Analysis Project Summary\n")
    f.write("=====================================\n\n")
    f.write(f"Original shape: {str((920, 22))}\n")
    f.write(f"Cleaned shape: {df.shape}\n")
    f.write(f"Features used: {X.shape[1]}\n")
    f.write(f"Target distribution:\n{y.value_counts().to_string()}\n\n")
    f.write("Leakage note:\n")
    f.write("The variable 'num' was excluded from model inputs because it directly maps onto the target label.\n\n")
    f.write("PCA explained variance ratio:\n")
    f.write(str(pca.explained_variance_ratio_))
    f.write("\n\nModel comparison:\n")
    f.write(results_df.to_string(index=False))

print("\nDone. All files are saved inside the 'outputs' folder.")
```
